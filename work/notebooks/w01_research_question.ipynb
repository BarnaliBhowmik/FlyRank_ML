## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

**My response:**
The numbers in the cell below make Lane 4 worth investigating bbbecause 9,579 of the 30,000 starter rows (32.53%) meet the initial low-CTR visibility rule. This gives  me a substantial set of potential review candidates rather than a very small or sparse problem. However, these are only candidates; low CTR doesn't by itself establish that a page needs a metadata or content change.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("/content/content_refresh_anonymized.csv")

print("Dataset shape:", df.shape)
print("Missing values in Lane 4 fields:")
print(
    df[
        [
            "impressions_90d",
            "clicks_90d",
            "ctr",
            "avg_position",
            "sessions_90d",
            "engagement_rate",
            "scroll_rate"
        ]
    ].isna().sum()
)
```

```python
visible_low_ctr = df[
    (df["impressions_90d"] >= 500) &
    (df["avg_position"] > 0) &
    (df["avg_position"] <= 20) &
    (df["ctr"] < 0.5)
]

print("Potential low-CTR review candidates:", len(visible_low_ctr))
print(
    "Percentage of dataset:",
    round(len(visible_low_ctr) / len(df) * 100, 2),
    "%"
)
```
