## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**My response:**
A fixed if-statement is too messy/limited as CTR and engagement depend on multiple factors like impressions, search position, sessions and engagement. The same CTR can mean different things for pages with different levels of visibility or search position, so one threshold cannot capture all these combinations or reliably rank review priority. A scoring approach can combine these signals and produce a more useful ranking. I will only claim that ML is better if it improves the ranking metric compared with the fixed-rule baseline.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

baseline_candidates = df[
    (df["impressions_90d"] >= 500) &
    (df["avg_position"] > 0) &
    (df["avg_position"] <= 20) &
    (df["ctr"] < 0.5)
]

print("Fixed-rule baseline candidates:", len(baseline_candidates))
print(
    "Baseline candidate rate:",
    round(len(baseline_candidates) / len(df) * 100, 2),
    "%"
)
```
